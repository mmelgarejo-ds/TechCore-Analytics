### 2. Tablas de dimensión — Ciudades · Sucursales · Vendedores

```python
# Creamos la tabla de Ciudades a partir de valores únicos
Ciudades = df['CiudadSucursal'].unique()
Ciudades = pd.DataFrame(Ciudades, columns=['Ciudad'])
# Asignamos un ID único a cada ciudad 
Ciudades['CiudadID'] = Ciudades.index + 1
# Reorganizamos las columnas
Ciudades = Ciudades[['CiudadID', 'Ciudad']]
Ciudades.head()
```

```python
# Creamos la tabla de Sucursales
Sucursales = df[['SucursalNombre', 'CiudadSucursal']].drop_duplicates().reset_index(drop=True) #eliminamos filas repetidas para obtener sucursales únicas
Sucursales['SucursalID'] = Sucursales.index + 1
# Merge con la tabla Ciudades para obtener el CiudadID
Sucursales = Sucursales.merge(Ciudades, left_on='CiudadSucursal', right_on='Ciudad', how='left')
Sucursales = Sucursales[['SucursalID', 'SucursalNombre', 'CiudadID']]
Sucursales
```

```python
# Creamos la tabla de Vendedores
Vendedores = df['VendedorNombre'].unique()
Vendedores = pd.DataFrame(Vendedores, columns=['VendedorNombre'])
# Asignamos un ID a cada vendedor
Vendedores['VendedorID'] = Vendedores.index + 1
Vendedores = Vendedores[['VendedorID', 'VendedorNombre']]
Vendedores.head()
```

```python
print(f"Total vendedores: {len(Vendedores)}")
```

### 3. Tabla Clientes — con validación de duplicados por nombre

```python
# Creamos la tabla de Clientes
Clientes = df[['ClienteNombre', 'GeneroCliente', 'EdadCliente', 'EmailCliente',
               'TelefonoCliente', 'DireccionCliente']].drop_duplicates().reset_index(drop=True)# Eliminamos duplicados
# Generamos un ClienteID
Clientes['ClienteID'] = Clientes.index + 1
Clientes = Clientes[['ClienteID', 'ClienteNombre', 'GeneroCliente', 'EdadCliente', 'EmailCliente',
                      'TelefonoCliente', 'DireccionCliente']]
Clientes.head()
```

### 4. Tabla MetodosPago

```python
# Creamos la tabla de Métodos de Pago
MetodosPago = df['MetodoPago'].dropna().unique() #eliminamos valores nulos
MetodosPago = pd.DataFrame(MetodosPago, columns=['MetodoPago'])
MetodosPago['MetodoPagoID'] = MetodosPago.index + 1
MetodosPago = MetodosPago[['MetodoPagoID', 'MetodoPago']]
MetodosPago
```

### 5. Tabla Facturas — tabla central del modelo

```python
# Verificamos el número total de filas
len(df)
```

```python
# Creamos la tabla de Facturas
Facturas = df[['VentaID', 'FechaVenta', 'Año', 'Mes', 'HoraVenta', 'SucursalNombre', 'VendedorNombre',
               'ClienteNombre', 'EmailCliente', 'MetodoPago', 'DescuentoVenta', 'TotalVenta']]

#Múltiples merge para reemplazar nombres con IDs
Facturas = Facturas.merge(Sucursales, left_on='SucursalNombre', right_on='SucursalNombre', how='left')
Facturas = Facturas.merge(Vendedores, left_on='VendedorNombre', right_on='VendedorNombre', how='left')
Facturas = Facturas.merge(Clientes, left_on=['ClienteNombre', 'EmailCliente'], right_on=['ClienteNombre', 'EmailCliente'], how='left')
Facturas = Facturas.merge(MetodosPago, left_on='MetodoPago', right_on='MetodoPago', how='left')

# Seleccionamos solo las columnas finales con los IDs
Facturas = Facturas[['VentaID', 'FechaVenta', 'Año', 'Mes', 'HoraVenta', 'SucursalID', 'VendedorID',
                      'ClienteID', 'MetodoPagoID', 'DescuentoVenta', 'TotalVenta']]
Facturas.head()

print(Facturas.columns.tolist())
print(Facturas['FechaVenta'].head())
```

```python
print("Filas en Facturas:", len(Facturas))
print("¿Hay VentaID repetidos?", Facturas['VentaID'].duplicated().sum())
print("Nulos por columna:")
print(Facturas.isnull().sum())
```

---
### 6. Tabla Productos — catálogo único a partir de 3 columnas de producto

```python
# Creamos la tabla de Productos
Productos = pd.DataFrame(columns=['NombreProducto', 'MarcaProducto', 'Precio'])

# Iteramos sobre los 3 productos posibles
for x in range(1, 4):
    append_product = pd.DataFrame(columns=['NombreProducto', 'MarcaProducto', 'Precio'])
    # Extraemos la información de cada producto
    append_product['NombreProducto'] = df[f'NombreProducto{x}']
    append_product['MarcaProducto'] = df[f'MarcaProducto{x}']
    append_product['Precio'] = df[f'PrecioUnitarioProducto{x}']
    # Concatenamos los productos al DataFrame principal
    Productos = pd.concat([Productos, append_product], ignore_index=True)

# Eliminamos duplicados y valores nulos para obtener productos únicos
Productos = Productos.dropna().drop_duplicates().reset_index(drop=True)
Productos['ProductoID'] = Productos.index + 1
Productos = Productos[['ProductoID', 'NombreProducto', 'MarcaProducto', 'Precio']]
Productos.head()
```

```python
print(f"Total productos únicos: {len(Productos)}")
```

---
### 7. Tabla DetalleFacturas — una fila por producto vendido

```python
# Creamos la tabla de Detalles de Facturas
DetalleFacturas = pd.DataFrame(columns=['VentaID', 'ProductoID', 'Cantidad', 'Subtotal'])

# Iteramos sobre los 3 posibles productos por venta
for i in range(1, 4):
    # Creamos un DataFrame temporal con los detalles de cada producto
    detalle_temp = df[['VentaID', f'NombreProducto{i}', f'MarcaProducto{i}', f'CantidadProducto{i}', f'SubtotalProducto{i}']]
    # Merge con la tabla Productos para obtener el ProductoID
    detalle_temp = detalle_temp.merge(Productos, left_on=[f'NombreProducto{i}', f'MarcaProducto{i}'], right_on=['NombreProducto', 'MarcaProducto'], how='left')
    # Seleccionamos solo las columnas necesarias
    detalle_temp = detalle_temp[['VentaID', 'ProductoID', f'CantidadProducto{i}', f'SubtotalProducto{i}']]
    # Renombramos las columnas para que sean consistentes
    detalle_temp = detalle_temp.rename(columns={f'CantidadProducto{i}': 'Cantidad', f'SubtotalProducto{i}': 'Subtotal'})
    # Concatenamos al DataFrame principal
    DetalleFacturas = pd.concat([DetalleFacturas, detalle_temp], ignore_index=True)

# Limpiamos los datos
DetalleFacturas = DetalleFacturas.dropna().reset_index(drop=True)
DetalleFacturas['Cantidad'] = DetalleFacturas['Cantidad'].astype(int)
DetalleFacturas['Subtotal'] = DetalleFacturas['Subtotal'].astype(float)
DetalleFacturas['VentaID'] = DetalleFacturas['VentaID'].astype(int)
DetalleFacturas['ProductoID'] = DetalleFacturas['ProductoID'].astype(int)
# Ordenamos por VentaID
DetalleFacturas.sort_values(by=['VentaID'], inplace=True)
DetalleFacturas.reset_index(drop=True, inplace=True)
# Creamos un ID único
DetalleFacturas['DetalleID'] = DetalleFacturas.index + 1
DetalleFacturas = DetalleFacturas[['DetalleID', 'VentaID', 'ProductoID', 'Cantidad', 'Subtotal']]
DetalleFacturas.head(20)
```

```python
print("Filas en DetalleFacturas:", len(DetalleFacturas))
print("ProductoID nulos:", DetalleFacturas['ProductoID'].isnull().sum())
```

### 9. Exportar modelo relacional → `modeloVentas.xlsx`

```python
# Exportamos todas las tablas a un archivo Excel
# Creamos una lista DataFrames a exportar
dataframes = [
    Ciudades,
    Sucursales,
    Vendedores,
    Clientes,
    MetodosPago,
    Productos,
    Facturas,
    DetalleFacturas
]

# Creamos una lista Hojas en Excel
nombres_hojas = [
    'Ciudades',
    'Sucursales',
    'Vendedores',
    'Clientes',
    'MetodosPago',
    'Productos',
    'Facturas',
    'DetalleFacturas'
]

# Reemplazamos valores nulos (NaN) con espacios en blanco
for df_tabla, nombre_hoja in zip(dataframes, nombres_hojas):
    df_tabla.fillna(' ', inplace=True)

# Ruta y nombre del archivo Excel
ruta_del_archivo = r"C:\Users\mau93\OneDrive\Desktop\Data Science\ProyectoM3_Mauricio_Melgarejo\Avances\Avance_2\modeloVentas.xlsx"

# Usamos ExcelWriter para escribir múltiples hojas en un solo archivo
with pd.ExcelWriter(ruta_del_archivo, engine='xlsxwriter') as writer:
    for nombre_hoja, df_tabla in zip(nombres_hojas, dataframes):
        df_tabla.to_excel(writer, sheet_name=nombre_hoja, index=False)

print(f"Archivo exportado: {ruta_del_archivo}")
```
